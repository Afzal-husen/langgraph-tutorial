```python
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, START, END
```

```python
class EquationState(TypedDict):
    a: int
    b: int
    c: int
    equation: str
    discriminant: int
    result: str
```

```python
def show_equation(state: EquationState):
   equation =  f"{state['a']}x2 + {state['b']}x + {state['c']}"
   state['equation'] = equation
   return state
```

```python
def calculate_discriminant(state: EquationState):
    discriminant =  state['b']**2 - 4 * state['a'] * state['c']
    state['discriminant'] = discriminant
    return state
```

```python
def real_roots(state:EquationState):
    root1 = (- state['b'] + state['discriminant'] ** 0.5) / 2 * state['a']
    root2 = (- state['b'] - state['discriminant'] ** 0.5) / 2 * state['a']
    result = f"Roots are {root1} and {root2}"
    state['result'] = result
    return state
```

```python
def repeat_roots(state: EquationState):
    root = - state['b']  / 2 * state['a']
    result = f"Roots are {root} and {root}"
    state['result'] = result
    return state
```

```python
def no_real_roots(state: EquationState):
    result = f"No real roots"
    state['result'] = result
    return state
```

```python
def check_condition(state: EquationState) -> Literal['real_roots', 'repeat_roots', 'no_real_roots']:
    discriminant = state['discriminant']
    if discriminant > 0:
        return 'real_roots'
    elif discriminant < 0:
        return "no_real_roots"
    else:
        return "repeat_roots"
        
```

```python
graph = StateGraph(EquationState)

graph.add_node('show_equation', show_equation)
graph.add_node('calculate_discriminant', calculate_discriminant)
graph.add_node('real_roots', real_roots)
graph.add_node('repeat_roots', repeat_roots)
graph.add_node('no_real_roots', no_real_roots)

graph.add_edge(START, 'show_equation')
graph.add_edge('show_equation', 'calculate_discriminant')
graph.add_conditional_edges('calculate_discriminant', check_condition)
graph.add_edge('real_roots', END)
graph.add_edge('repeat_roots', END)
graph.add_edge('no_real_roots', END)

workflow = graph.compile()
```

```python
initial_state = {
    "a": 4,
    "b": -5,
    "c": -4
}
workflow.invoke(initial_state)
```
